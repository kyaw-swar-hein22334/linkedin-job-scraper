```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
import time
import pandas as pd

# --- LinkedIn URL for Data Analyst jobs in Berlin ---
url = "https://www.linkedin.com/jobs/search/?keywords=Data%20Analyst&location=Berlin%2C%20Germany"

# --- Start Driver ---
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))
driver.maximize_window()
driver.get(url)
time.sleep(5)

# --- Keep scrolling and clicking 'See more jobs' until no new jobs ---
last_height = driver.execute_script("return document.body.scrollHeight")

while True:
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(3)

    # Try clicking "See more jobs" button
    try:
        see_more = driver.find_element(By.XPATH, "//button[@aria-label='See more jobs']")
        driver.execute_script("arguments[0].click();", see_more)
        print("🔽 Clicked 'See more jobs'")
        time.sleep(3)
    except:
        pass

    # Check if page height stopped growing
    new_height = driver.execute_script("return document.body.scrollHeight")
    if new_height == last_height:
        print("✅ Reached end of jobs list")
        break
    last_height = new_height

# --- Now scrape all job cards ---
titles, companies, locations, links = [], [], [], []

jobs = driver.find_elements(By.CLASS_NAME, "base-card")
print(f"📄 Total jobs loaded: {len(jobs)}")

for idx, job in enumerate(jobs, start=1):
    try:
        title = job.find_element(By.CLASS_NAME, "base-search-card__title").text
    except:
        title = None
    try:
        company = job.find_element(By.CLASS_NAME, "base-search-card__subtitle").text
    except:
        company = None
    try:
        location = job.find_element(By.CLASS_NAME, "job-search-card__location").text
    except:
        location = None
    try:
        link = job.find_element(By.CSS_SELECTOR, "a.base-card__full-link").get_attribute("href")
    except:
        link = None

    titles.append(title)
    companies.append(company)
    locations.append(location)
    links.append(link)

    # --- Progress counter ---
    if idx % 50 == 0 or idx == len(jobs):  # print every 50 jobs, and final one
        print(f"✅ Scraped {idx}/{len(jobs)} jobs")

# --- Save results ---
df = pd.DataFrame({
    "Title": titles,
    "Company": companies,
    "Location": locations,
    "Link": links
})

df.to_csv("linkedin_jobs_berlin.csv", index=False, encoding="utf-8")
print(f"🎉 Scraping finished! {len(df)} jobs saved to linkedin_jobs_berlin.csv")

driver.quit()
```
